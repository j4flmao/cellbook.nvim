# cellbook.nvim demo

Run a cell with `<S-CR>` (or `<localleader>r`), run everything with `<localleader>R`.

```python
import sys
print(f'Python {sys.version.split()[0]}')
6 * 7
```

## Streaming output

Keep typing while this runs.

```python
import time
for i in range(5):
    print(f'step {i}', flush=True)
    time.sleep(0.4)
```

## Errors

Tracebacks keep their colours.
